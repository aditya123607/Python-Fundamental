```python
# Linear Probing in Hash Table

size = 10
hash_table = [None] * size

keys = [23, 43, 13, 27, 37]

for key in keys:
    index = key % size

    # Linear probing
    while hash_table[index] is not None:
        index = (index + 1) % size

    hash_table[index] = key

# Display hash table
print("Hash Table for Assignment 1:")

for i in range(size):
    print(i, "->", hash_table[i])
```
