```python
# Import Deque from collections module
from collections import deque

#create a graph using an adjacency list representation
#each key is a vertex and its value is a list of adjacent vertices
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'E'],
    'E': ['C', 'D'],
}

# START  = the vertex from which the BFS traversal will begin
def bfs(graph, start):

    # create an empty set to store vertex that have already visited 
    visited = set()

    #create an empty queue
    # BFS follows FIFO 
    queue = deque()

    # here start = 'A' , so A became visited 
    visited.add(start)

    # Add the starting vertex to the queue 
    queue.append(start)

    # continue the loop as long as th equeue is not empty
    while queue:
     # remove the first vertex from the queue.
     # popleft() remove from the front of the queue
        vertex = queue.popleft()

        # print the vertex that is jusu removed.
        # ends = "" keep the output on the same line.
        print(vertex, end=" ")


        # get all neibour of the current vertex and pross them oe by one 
        for neighbor in graph[vertex]:

            # check whether this neibour has onlu ben visited or not
            if neighbor not in visited:

                # mark the neiboyr as visited
                # this presenrt visiting the same vertex again 
                visited.add(neighbor)

                #add the unvisited neighbour to the queue so it will procees later.
                queue.append(neighbor)

# call the BFS functiion with the graph and starting vertex 'A'
bfs(graph, 'A')


```
