```python
# Q.1  Write a function using the two-pointer technique to check whether a given array is a palindrome.

def is_palindrome(arr):
    left = 0
    right = len(arr) - 1
    
    while left < right:
        if arr[left] != arr[right]:
            return False
        left += 1
        right -= 1
        
    return True

print(is_palindrome([1, 2, 3, 2, 1]))  
print(is_palindrome([1, 2, 3, 4, 5]))  
```

```python
 #  Q.2 Given a sorted array of integers, write a function using two pointers to find a pair of elements whose sum equals a given target value.

def find_pair_indices_with_sum(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            return (left, right)
        elif current_sum < target:
            left += 1
        else:
            right -= 1

    return None
```

```python
# Q.3 Write a program using the two-pointer approach to reverse an array in place (without using extra space).

def reverse_array_in_place(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]

        left += 1
        right -= 1
numbers = [1, 2, 3, 4, 5, 6]
print("Original Array:", numbers)

reverse_array_in_place(numbers)
print("Reversed Array:", numbers)
```

```python
 # Q.4 Given an array containing both positive and negative numbers, use the two-pointer method to segregate negative and positive numbers on either side.

def segregate_pos_neg(arr):
    left = 0
    right = len(arr) - 1

    while left <= right:
        if arr[left] < 0:
            left += 1
        elif arr[right] >= 0:
            right -= 1
        else:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1

    return arr

numbers = [-12, 11, -13, -5, 6, -7, 5, -3, -6]
print("Original Array:", numbers)

segregate_pos_neg(numbers)
print("Segregated Array:", numbers)
```

```python
 # Q.5 Using two pointers, write a function to remove duplicates from a sorted array in place and return the new length.

def remove_duplicates(nums):
    if not nums:
        return 0

    write_index = 1
    for read_index in range(1, len(nums)):
        if nums[read_index] != nums[read_index - 1]:
            nums[write_index] = nums[read_index]
            write_index += 1

    return write_index
numbers = [1, 1, 2, 2, 3, 4, 4, 5]
print("Original Array:", numbers)

new_length = remove_duplicates(numbers)
print("Length after removing duplicates:", new_length)
print("Modified Array (first k elements):", numbers[:new_length])
```

```python
# Q.6 write a user defines functionto find the factorial of a number using recursion 

def factorial(n: int) -> int:
    # Base case: factorial of 0 or 1 is 1
    if n <= 1:
        return 1

    # Recursive case: n! = n * (n - 1)!
    return n * factorial(n - 1)

number = 5
result = factorial(number)
print(f"The factorial of {number} is {result}")
```
