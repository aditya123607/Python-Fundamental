```python
class Car:
    # Use __init__ to initialize the object attributes
    def __init__(self, brand, color, top_speed):
        self.brand = brand
        self.color = color
        self.top_speed = top_speed
        self.current_speed = 0

    # Indent the method so it sits inside the Car class
    def accelerate(self):
        self.current_speed += 20
        print(f"The {self.color} {self.brand} is now moving at {self.current_speed} km/h")

# Creating objects from the class (Instantiating)
car1 = Car("Audi", "red", 354)
car2 = Car("Toyota", "blue", 240)

# Testing the method
car1.accelerate()
car2.accelerate()
```
