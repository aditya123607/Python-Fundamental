```python
# Q.1 Write a function using the two-pointer technique to check whether a given array is a palindrome.
nums = [2, 4, 6, 8, 6, 4, 2]

left = 0
right = len(nums) - 1

while left < right:
    if  nums[left] != nums[right]:
        print("Not palindrome")
        break

    left += 1
    right -= 1

else:
    print("Palindrome")
```

```python
# Q.2 Given a sorted array of integers, write a function using two pointers to find a pair of elements whose sum equals a given target value.

nums = [1,3,5,8,9]

target = 11
left = 0
right = len(nums)-1

while left < right:
    result = nums[left] + nums[right] 
    if result == target:
        print("index position:", left, right)
        print("index position element:" , nums[left], nums[right])
        print("targeted sum :", result)
        break
    elif result < target:
        left +=1
    else:
        right -=1
```

```python
# Q.3 Write a program using the two-pointer approach to reverse an array in place (without using extra space).
nums =[1,2,5,4,6,9,7,8,3]

left =0
right = len(nums) -1

while left < right:
    nums[left] , nums[right] = nums[right], nums[left]
    left +=1
    right -=1
print(nums)
```

```python
#Q4. Given an array containing both positive and negative numbers, use the two-pointer method to segregate negative and positive numbers on either side.
nums = [-1,-5,2,-3,5,8,6]

m=0
for i in range(len(nums)):
    if nums[i] < 0:
        nums[m], nums[i] = nums[i], nums[m]
        m +=1

print(nums)
```

```python
#Q5. Using two pointers, write a function to remove duplicates from a sorted array in place and return the new length.

def remove_duplicates(nums):
    i = 0
    for j in range(1, len(nums)):
        if nums[j] != nums[i]:
            i +=1
            nums[i] = nums[j]

    return i+1
nums=[0,1,2,2,3,3,5,5,6]
result = remove_duplicates(nums)
print(result)
print(nums[:result])
```

```python
#Q6. Write a user-defined function int factorial(int n) that calculates and returns the factorial of a given number using recursion.

def factorial(n):
    if n == 0 or n== 1:
        return 1
    return n * factorial(n - 1)  #using recursion

print(factorial(5))
```

```python
#Q7. Define a function bool isPrime(int n) that checks whether a given number is prime, and use it inside main() to print all prime numbers between 1 and 50.

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True

def main():
    print("Prime numbers between 1 and 50")
    for num in range(1,51):
        if is_prime(num):
            print(num, end =" ")
    print()

main()
```

```python
#Q9. Create a function int gcd(int a, int b) to find the GCD of two numbers using the Euclidean algorithm, and call it from main().
def gcd(a , b):
    while b != 0:
        a , b = b , a%b
        return b

def main():
    num1 = 36 
    num2 = 24

    result = gcd(num1, num2)
    print(f"GCD of {num1} and {num2} is {result}")

main()
```

```python
 #Q10. write a user-defined function float average(int arr[],int size) that takes an array of integers and its size as input and returns the average of the elements in the array as a float.
 
 def average(arr, size):
    while size != 0:
        total_sum = sum(arr)
        return float(total_sum/size)

nums =[10,20,30,40,50,60]
size = len(nums)
result = average(nums, size)

print(result)
```
